## **Теоретичні відомості**

**Часовий ряд** — це впорядкована послідовність даних, що зібрані в різні моменти часу. Прикладами таких даних є щоденні курси валют, щомісячні продажі товарів, річні температури чи обсяги виробництва. Аналіз часових рядів дозволяє виявити закономірності та тенденції, а також прогнозувати майбутні значення на основі історичних даних.

**Сфери застосування часових рядів**

Часові ряди широко використовуються в різних галузях:

* Економіка: аналіз ринкових цін, прогнозування попиту та обсягу продажів.
* Метеорологія: прогноз погоди та кліматичних умов.
* Інженерія: моніторинг виробничих процесів, виявлення збоїв.
* Фінанси: моделювання курсу акцій та валют.
* Соціологія: аналіз демографічних даних та трендів.

### **Компоненти часових рядів**


Часовий ряд може складатися з кількох основних компонентів:

1. Тренд (Trend) – довгострокова тенденція, яка відображає загальний напрямок
зміни даних (зростання або спад).
2. Сезонність (Seasonality) – регулярні коливання, які повторюються через певний період (наприклад, річні або щомісячні коливання).
3. Циклічність (Cyclical) – довгострокові коливання, що не мають регулярного періоду, але можуть бути пов'язані з економічними циклами.
4. Шум (Noise) – випадкові коливання, що не піддаються прогнозуванню та не мають закономірностей.

### **Методи аналізу часових рядів**

Для роботи з часовими рядами використовують різні методи:

1. Скользяче середнє (Moving Average) – метод згладжування даних, який допомагає зменшити вплив випадкових коливань.

   * Просте скользяче середнє: обчислюється як середнє значення за певний період.
  * Лінійно-зважене скользяче середнє: кожному значенню надається своя вага, яка змінюється з часом.
2. Експоненційне згладжування (Exponential Smoothing) – метод, при якому більшій ваги надаються останнім спостереженням, що дозволяє краще прогнозувати майбутні значення.

3. Декомпозиція часових рядів – процес розкладу ряду на окремі компоненти (тренд, сезонність, шум) для кращого аналізу.

4. Автокореляція (Autocorrelation) – оцінка залежності значень ряду від попередніх спостережень. Використовується для виявлення патернів у часі.



### **Моделі для прогнозування часових рядів**

1. ARIMA (Auto-Regressive Integrated Moving Average) – одна з найпопулярніших моделей для прогнозування часових рядів. Вона поєднує в собі авторегресію (AR), інтегровану частину (I) для роботи з нестаціонарними рядами та скользяче середнє (MA).

2. Holt-Winters – метод експоненційного згладжування, який добре підходить для рядів із сезонністю. Має два варіанти:

3. Адитивний (для рядів із постійною амплітудою сезонних коливань).
Мультиплікативний (для рядів із сезонністю, яка змінюється з часом).
4. Prophet – сучасний інструмент від Facebook для прогнозування часових рядів, який автоматично враховує тренди та сезонні компоненти. Зручний для роботи з великими наборами даних та гнучкий у налаштуваннях.

### **Оцінка якості прогнозу**

Для оцінки точності прогнозу використовуються такі метрики:

* MAE (Mean Absolute Error) – середня абсолютна помилка.
* RMSE (Root Mean Squared Error) – корінь середньої квадратичної помилки.
* MAPE (Mean Absolute Percentage Error) – середня абсолютна відносна помилка у відсотках.

Ці метрики дозволяють порівняти результати різних моделей і вибрати найкращу для конкретних даних.
